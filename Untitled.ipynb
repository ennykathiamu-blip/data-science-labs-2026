```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

# Different seed -> different random data than the original notebook
np.random.seed(7)

# Independent variable: training hours per employee per quarter (0-40 hours)
X = 40 * np.random.rand(150, 1)

# Dependent variable: productivity score, following a diminishing-returns curve (sqrt),
# NOT a straight line like the original example
y = 50 + 25 * np.sqrt(X) + 6 * np.random.randn(150, 1)

df = pd.DataFrame({'Training_Hours': X.flatten(), 'Productivity_Score': y.flatten()})

print("Synthetic Data Generated (first 5 rows):")
print(df.head())
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Training_Hours'], df['Productivity_Score'], alpha=0.6, color='teal')
plt.title('Training Hours vs. Productivity Score (Synthetic Data)')
plt.xlabel('Training Hours per Quarter')
plt.ylabel('Productivity Score')
plt.grid(True)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=7)

# Model A: plain linear regression
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)
r2_linear = r2_score(y_test, y_pred_linear)

# Model B: polynomial regression (degree 2) to capture the curve
poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)
y_pred_poly = poly_model.predict(poly.transform(X_test))
r2_poly = r2_score(y_test, y_pred_poly)

print(f"Linear model R-squared:     {r2_linear:.3f}")
print(f"Polynomial model R-squared: {r2_poly:.3f}")
```

```python
hours_grid = np.linspace(0, 40, 200).reshape(-1, 1)
linear_curve = linear_model.predict(hours_grid)
poly_curve = poly_model.predict(poly.transform(hours_grid))

plt.figure(figsize=(10, 6))
plt.scatter(X_test, y_test, alpha=0.5, label='Actual (Test Data)', color='gray')
plt.plot(hours_grid, linear_curve, color='red', linewidth=2, label=f'Linear fit (R2={r2_linear:.2f})')
plt.plot(hours_grid, poly_curve, color='green', linewidth=2, label=f'Polynomial fit (R2={r2_poly:.2f})')
plt.title('Training Hours vs. Productivity: Linear vs. Polynomial Fit')
plt.xlabel('Training Hours per Quarter')
plt.ylabel('Productivity Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
a, b = poly_model.coef_[0]
print(f"Polynomial coefficients -> a (linear term): {a:.3f}, b (quadratic term): {b:.3f}")
print(f"Intercept: {poly_model.intercept_[0]:.3f}")
print()
print("Marginal productivity gain per additional hour, at different training levels:")
for h in [5, 10, 20, 30, 40]:
    slope = a + 2 * b * h
    print(f"  At {h:>2} hours: +{slope:.2f} productivity points per additional hour")
```
